# Fir dashboard
Tap **Run → Run All Cells**. After ~20 s a blue **Open dashboard in a new tab** button appears, with the dashboard below it.


```python
import os, sys, html
from pathlib import Path
sys.path.insert(0, "/project/def-kanj241/hvt632/fir_interface")
from firdash.render import build
from IPython.display import HTML

DAYS = 7  # job-history window

page = build({"days": str(DAYS), "static": "1"})
out = Path.home() / "fir_dashboard.html"
out.write_text(page, encoding="utf-8")
url = os.environ.get("JUPYTERHUB_SERVICE_PREFIX", "/") + "files/fir_dashboard.html"
HTML(f'<a href="{url}" target="_blank" style="display:inline-block;margin:6px 0 14px;padding:12px 22px;'
     'background:#3987e5;color:#fff;border-radius:8px;font:600 16px system-ui;text-decoration:none">'
     'Open dashboard in a new tab &#8599;</a>'
     f'<iframe srcdoc="{html.escape(page)}" style="width:100%;height:1700px;border:0;border-radius:8px" '
     'onload="this.style.height=(this.contentDocument.documentElement.scrollHeight+20)+\'px\'"></iframe>')
```
